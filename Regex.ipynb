```python
import re

month_map = {
    "فروردین": 1, "اردیبهشت": 2, "خرداد": 3, "تیر": 4,
    "مرداد": 5, "شهریور": 6, "مهر": 7, "آبان": 8,
    "آذر": 9, "دی": 10, "بهمن": 11, "اسفند": 12
}

def fa_to_en_num(s: str) -> str:
    trans = str.maketrans("۰۱۲۳۴۵۶۷۸۹", "0123456789")
    return s.translate(trans)

day_words = {
    "یکم": 1, "اول": 1, "دوم": 2, "سوم": 3, "چهارم": 4,
    "پنجم": 5, "ششم": 6, "هفتم": 7, "هشتم": 8, "نهم": 9, "دهم": 10,
    "یازدهم": 11, "دوازدهم": 12, "سیزدهم": 13, "چهاردهم": 14, "پانزدهم": 15,
    "شانزدهم": 16, "هفدهم": 17, "هجدهم": 18, "نوزدهم": 19, "بیستم": 20,
    "بیست و یکم": 21, "بیست و دوم": 22, "بیست و سوم": 23, "بیست و چهارم": 24, "بیست و پنجم": 25,
    "بیست و ششم": 26, "بیست و هفتم": 27, "بیست و هشتم": 28, "بیست و نهم": 29, "سی‌ام": 30, "سی ام": 30,
    "سی و یکم": 31
}

months = "|".join(month_map.keys())
patterns = [
    rf"(?P<y1>[0-9۰-۹]{{4}})[/-](?P<m1>[0-9۰-۹]{{2}})[/-](?P<d1>[0-9۰-۹]{{2}})",
    rf"(?P<d2>[0-9۰-۹]{{1,2}})[/-](?P<mn2>{months})[/-](?P<y2>[0-9۰-۹]{{4}})",
    rf"(?P<dw>{'|'.join(sorted(day_words.keys(), key=len, reverse=True))})\s+(?P<mn3>{months})(?:\s*ماه)?(?:\s*سال)?\s*(?P<y3>[0-9۰-۹]{{4}})"
]

big_pattern = re.compile("|".join(patterns))

def extract_dates(text):
    results = []
    for m in big_pattern.finditer(text):
        if m.group("y1"):
            y, mth, d = map(lambda x: int(fa_to_en_num(x)), [m.group("y1"), m.group("m1"), m.group("d1")])
        elif m.group("y2"):
            d = int(fa_to_en_num(m.group("d2")))
            mth = month_map[m.group("mn2")]
            y = int(fa_to_en_num(m.group("y2")))
        elif m.group("y3"):
            d = day_words[m.group("dw")]
            mth = month_map[m.group("mn3")]
            y = int(fa_to_en_num(m.group("y3")))
        else:
            continue
        results.append((y, mth, d))
    return results


text = """
۱۴۰۳/۰۶/۲۵
۲۵/شهریور/۱۴۰۳
۱۴۰۳-۰۶-۰۵
بیست و پنجم شهریور ماه سال ۱۴۰۳
پانزدهم دی ماه سال ۱۴۰۳
یکم مرداد ماه سال ۱۳۲۶
"""

dates = extract_dates(text)
for y, m, d in dates:
    print(f"{y:04d}-{m:02d}-{d:02d}")
```

```python
import re

# عبارت منظم برای تشخیص یک شیء JSON با یک سطح تو در تو
# این عبارت برای JSONهای پیچیده و چندسطحی کار نمی‌کند
json_object_regex = r'\{\s*(?:"[^"]*"\s*:\s*(?:"[^"]*"|\d+(?:\.\d+)?|true|false|null|\[.*?\]|\{.*?\}))(?:\s*,\s*"[^"]*"\s*:\s*(?:"[^"]*"|\d+(?:\.\d+)?|true|false|null|\[.*?\]|\{.*?\}))*\s*\}'

# مثال‌ها برای تست
test_objects = [
    '{"name": "Alice", "age": 30}',
    '{"city": "Tehran", "isCapital": true, "population": 8.7}',
    '{"data": ["item1", "item2"], "metadata": {"status": "ok"}}',
    '{"key": "value"}',
    '{"a": 1, "b": {"c": 2}}'
]

# رشته‌های نامعتبر
invalid_objects = [
    '{"key": "value"',
    '{"key": "value" "other_key": "other_value"}',
    '{"key": "value",}',
    '{"key": "value", "key2": ["a", "b", "c',
    '{"a": 1, "b": {"c": 2, "d": [3, 4]}}',
    '{"key1": {"key2": {"key3": "value"}}}'# این مثال به دلیل دو سطح تو در تو fail می‌شود
]

print("--- تست شیء‌های معتبر ---")
for s in test_objects:
    if re.fullmatch(json_object_regex, s):
        print(f"'{s}' -> معتبر است.")
    else:
        print(f"'{s}' -> نامعتبر است.")

print("\n--- تست شیء‌های نامعتبر ---")
for s in invalid_objects:
    if re.fullmatch(json_object_regex, s):
        print(f"'{s}' -> معتبر است.")
    else:
        print(f"'{s}' -> نامعتبر است.")
```

```python
import re

# عبارت منظم برای تشخیص یک آرایه JSON با یک سطح تو در تو
# این عبارت برای JSONهای پیچیده و چندسطحی کار نمی‌کند
json_array_regex = r'\[\s*(?:(?:"[^"]*"|\d+(?:\.\d+)?|true|false|null|\[.*?\]|\{.*?\}))(?:\s*,\s*(?:"[^"]*"|\d+(?:\.\d+)?|true|false|null|\[.*?\]|\{.*?\}))*\s*\]'


# مثال‌ها برای تست
test_arrays = [
    '["apple", "banana", "cherry"]',
    '[1, 2, 3]',
    '[true, false, null]',
    '[1, "text", true, {"key": "value"}]',
    '[[1, 2], [3, 4]]',
]

# رشته‌های نامعتبر
invalid_arrays = [
    '["item1", "item2"',
    '[1, 2, 3,]',
    '["item1", {"key": "value"',
    '[["a"], ["b", ["c"]]]'  # این مثال به دلیل دو سطح تو در تو fail می‌شود
]

print("--- تست آرایه‌های معتبر ---")
for s in test_arrays:
    if re.fullmatch(json_array_regex, s):
        print(f"'{s}' -> معتبر است.")
    else:
        print(f"'{s}' -> نامعتبر است.")

print("\n--- تست آرایه‌های نامعتبر ---")
for s in invalid_arrays:
    if re.fullmatch(json_array_regex, s):
        print(f"'{s}' -> معتبر است.")
    else:
        print(f"'{s}' -> نامعتبر است.")
```

```python
import re

json_verbose_regex = re.compile(r"""
    (
        \{
        \s*
        (?:"[^"]*"\s*:\s*(?:"[^"]*"|\d+(?:\.\d+)?|true|false|null|\[.*?\]|\{.*?\}))
        (?:\s*,\s*"[^"]*"\s*:\s*(?:"[^"]*"|\d+(?:\.\d+)?|true|false|null|\[.*?\]|\{.*?\}))*
        \s*
        \}
        |
        \[
        \s*
        (?:(?:"[^"]*"|\d+(?:\.\d+)?|true|false|null|\[.*?\]|\{.*?\}))
        (?:\s*,\s*(?:"[^"]*"|\d+(?:\.\d+)?|true|false|false|null|\[.*?\]|\{.*?\}))*
        \s*
        \]
    )
""", re.VERBOSE)

test_cases = [
    '{"name": "Alice"}',
    '[1, 2, 3]',
    '{"data": ["item1", "item2"]}',
    '["item1", {"key": "value"}]',
    '{"name": "Alice"',
    '[1, 2, 3,',
    '{"a": {"b": {"c": 1}}}'
]

for s in test_cases:
    if json_verbose_regex.fullmatch(s):
        print(f"'{s}' -> معتبر ✅")
    else:
        print(f"'{s}' -> نامعتبر ❌")
```
